## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Content performance depends on many factors that interact with one another, including search volume, ranking position, CTR, content length, freshness, and user engagement. A simple rule such as "low CTR means declining" would ignore these interactions. Machine learning can discover complex patterns from historical data and make more reliable predictions than a fixed set of rules.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

features = [
    "search_volume",
    "word_count",
    "ctr",
    "avg_position",
    "engagement_rate",
    "content_age_days"
]

print(df[features].describe())
```
