## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Search Question**

Can we predict whether a webpage is likely to experience declining search performance using its SEO and traffic features?

**Decision**

The decision is which pages should receive attention first.

**Unit of Analysis**

One webpage.

Each row in the dataset represents a single webpage.

**Model Output**

A binary prediction:

Declining
Not Declining
Optionally, a probability score indicating how likely the page is to decline.

**Action**

If a page is predicted to decline, an SEO or content team could:

Update outdated content
Improve keyword targeting
Add internal links
Refresh metadata
Investigate technical SEO issues

**Cost of a Wrong Recommendation**

**False Positive**

A healthy page is predicted as declining.

Cost:

Time spent reviewing a page that did not need attention.

False Negative
A declining page is predicted as healthy.

Cost:

Lost traffic
Lower rankings
Reduced visibility
Missed business opportunities

False negatives are likely more costly because declining pages may continue losing traffic without intervention.

## 3. Quick look at the data (2-3 real numbers)

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nCorrelation between search_volume and impressions_90d:")
print(df["search_volume"].corr(df["impressions_90d"]))

print("\nCorrelation between avg_position and ctr:")
print(df["avg_position"].corr(df["ctr"]))

df["is_declining_label"] = (
    df["trend_direction"]
      .str.lower()
      .eq("down")
)

print("\nDeclining vs Non-declining pages:")
print(df["is_declining_label"].value_counts())
```
